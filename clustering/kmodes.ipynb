```python
import pandas as pd
from kmodes.kmodes import KModes
```

```python
df = pd.read_csv('kmodes_customer_preferences.csv')
```

```python
df.head()
```

```python
model = KModes(n_clusters=2)
model.fit_predict(df)
model.cost_
```

```python
cost = []
for cluster in range(1,6):
    model = KModes(n_clusters=cluster)
    model.fit_predict(df)
    cost.append(model.cost_)
```

```python
import matplotlib.pyplot as plt


plt.plot(range(1,6),cost,marker='o')
```
